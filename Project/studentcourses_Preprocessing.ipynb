```python
# import the needed libraries
import pandas as pd
```

```python
# Read all sheets from the Excel file
all_sheets = pd.read_excel('../Excel_sheets/studentcourses.xlsx', sheet_name=None)

all_sheets
```

```python
all_sheets.items()
```

```python
# Process each sheet and add the department column
for sheet_name, sheet_df in all_sheets.items():
    # Add new column at the beginning
    sheet_df.insert(0, 'القسم', sheet_name)
    
    # Clean the department name
    cleaned_name = sheet_name.strip()
    sheet_df['القسم'] = cleaned_name
    
    # Update the sheet in the dictionary
    all_sheets[sheet_name] = sheet_df
```

```python
# extract a single sheet from the sheets to check the data in sheets
first_sheet = list(all_sheets.values())[0]

# display the sheet
first_sheet
```

```python
# Drop the second column
for sheet_name, sheet_df in all_sheets.items():
    sheet_df.drop(sheet_df.columns[1], axis=1, inplace=True)
```

```python
# list of the names of the columns
col_names = ['القسم', 'رقم الطالب', 'اسم الطالب', 'التخصص', 'رقم المقرر', 'اسم المقرر', 'رمز المقرر', 'رقم الشعبة']

# change dataframe columns name
for sheet_name, sheet_df in all_sheets.items():
    if len(sheet_df.columns) == len(col_names):
        sheet_df.columns = col_names
```

```python
# concat all sheets in one sheet
all_sheets_df = pd.concat(all_sheets.values(), ignore_index=True)

all_sheets_df
```

```python
all_sheets_df.info()
```

```python
all_sheets_df.dropna(inplace= True)
```
